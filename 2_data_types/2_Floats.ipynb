# 🛟 Floats in Python
Floats represent decimal numbers like `3.14`, `0.001`, or `-7.5`.

## Why Floats Matter
You need floats whenever precision matters:

- Money calculations  
- Percentages  
- Distances  
- Scientific measurements  

```python
pi = 3.14159
temp = -5.7
price = 199.99

print(f"The value of 'pi' is {pi} and its type is {type(pi)}")
print(f"The value of 'temp' is {temp} and its type is {type(temp)}")
print(f"The value of 'price' is {price} and its type is {type(price)}")
```

## Common Float Operations
Floats support all arithmetic operations, plus:
- `round()`
- `abs()`
- `float()`

```python
x = 7.8
y = 2.1
z = 4

print(f"The sum of x and y is {x + y}.\n")
print(f"The subtraction of y from x gives {x - y}.\n")
print(f"x multipled by y is {x * y}.\n")
print(f"x divided by y is {x / y}.\n")
print(f"The quotent after dividing x by y is {x // y}.\n")
print(f"The remainder when you divide x by y is {x % y}.\n")
print(f"x power y is {x ** y}.\n")

print(f"x rounded off is {round(x)}.\n")
print(f"The absolute value of y is {abs(y)}.\n")

print(f"To change a value to a float, such as an integer or a stringified integer or null,you simply apply a float() function to the value. For example, {float(z)}")
```
