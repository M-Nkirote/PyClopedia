# Strings in Python
Strings represent text - characters, sentences, symbols, even emojis 💗

In Python, strings are enclosed in " ".

## Why Strings Matter
Strings appear everywhere:

- Usernames  
- Chat messages  
- File paths  
- Emails  
- API responses  
- Titles, labels, logs

```python
phrase_1 = "The sky is "
phrase_2 = "blue."


print(f"Phrase 1 is {phrase_1} and its of type {type(phrase_1)}.\n")
```

## Common String Operations
- Slicing: `s[0:3]`
- Length: `len(s)`
- Replace: `s.replace()`
- Upper/lower: `.upper()`, `.lower()`
- Split/join: `.split()`, `" ".join()`
- Formatting: f-strings  

```python
my_name = "Fridah"
```

```python
print(f"To capitalize all characters of a string, use a.upper() : {my_name.upper()}.")
```

```python
print(f"To have all characters of a string be lowercase, use a.lower() : {my_name.lower()}.")
```

```python
print(f"To join multiple strings, you can use '+', example phrase_1 + phrase_2: {phrase_1 + phrase_2}")
```

```python
print(f"To get certain characters of a string, you use slicing operations. For example: \n")

print(f"To get the first 3 characters of my_name : {my_name[0:3]}")

print(f"To get the last 2 characters of my_name : {my_name[-2:]}")

print(f"To get every second character in my name : {my_name[1::2]}")
```

```python
print(f"To split the charactes in a string you simply use the split() : {phrase_1.split(" ")}.")

# Here, it splits / breaks the phrase on a space character as indicated by the " "
```

```python
# You can use the join() to join each character of the string with a space between them.
# Here, as the user, you get to add a string in which a space will be added between each character

name = input("Enter a name: ")
name = " ".join(name)
print(name)
```

```python
# You can replace characters in your string values using replace()

print(f"To replace all instances of 'a' in 'Banana' with '🍌', you call the replace() & you get {"Banana".replace('a','🍌')}.")
```
