# 🧮 Integers in Python

Integers (`int`) are whole numbers — positive, negative, or zero.

They are one of the most common data types in programming.

## Why Integers Matter
Integers appear everywhere:

- Counting items  
- Looping through lists  
- Tracking scores  
- Indexing  
- Mathematical operations  
- Iterating over ranges  

Integers are the backbone of logic and computation.

```python
# Some examples 

a = 10
b = -3
c = 0

# (In this example, a,b and c are knows as variables and they acts as a placeholder or a container to 
# store some value. And in your operations, you can keep calling (using) such variables as they identify 
# the values. Thus, its important to choose meaningful variable names 😊)


print(f"The value of 'a' is {a} and it's of the type {type(a)}.")
print(f"The value of 'b' is {b} and it's of the type {type(b)}.")
print(f"The value of 'c' is {c} and it's of the type {type(c)}.")

# (In python, there are certain in-built functions such as print() and type() that have different functions.
# For example, print() is used when you want to display something on screen while type() is used to get the type of something.)
```

## Some of the most common integer operations are:

- Addition: `+`
- Subtraction: `-`
- Multiplication: `*`
- Division: `/`
- Floor division: `//`
- Modulus: `%` (remainder)
- Exponent: `**`


```python
x = 25
y = 4

print(f"The sum of x and y is {x + y}.")
print(f"The subtraction of y from x gives {x - y}.")
print(f"x multipled by y is {x * y}.")
print(f"x divided by y is {x / y}.")
print(f"The quotent after dividing x by y is {x // y}.")
print(f"The remainder when you divide x by y is {x % y}.")
print(f"x power y is {x ** y}.")
```

## Real-Life Use Cases
- Keeping track of quantities (items in a cart)
- Counting steps in loops
- Representing years, ages, floors, temperatures
- Game scores or life counters
